```python
import numpy as np, matplotlib.pyplot as plt
fs = 160
t = np.arange(0, 2, 1/fs)                        # 2 seconds
x = 20*np.sin(2*np.pi*10*t) + 8*np.sin(2*np.pi*22*t) + 5*np.random.randn(t.size)

X = np.fft.rfft(x)                               # FFT for real signals
freqs = np.fft.rfftfreq(x.size, d=1/fs)
amp = 2 * np.abs(X) / x.size                     # one-sided amplitude

plt.plot(freqs, amp); plt.xlim(0, 40)
plt.xlabel("Frequency (Hz)"); plt.ylabel("Amplitude (µV)"); plt.show()
```
