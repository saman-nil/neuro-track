```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
fs = 160                                  # same rate as your boss-fight data
t = np.arange(0, 1, 1/fs)                 # 1 second of time points
alpha = 20 * np.sin(2*np.pi*10*t)         # 10 Hz, 20 µV
noise = 7 * np.random.randn(t.size)      # random noise, 10 µV
eeg = alpha + noise
```

```python
plt.figure(figsize=(8, 3))
plt.plot(t, eeg)
plt.xlabel("Time (s)")
plt.ylabel("Amplitude (µV)")
plt.title("Fake occipital EEG: 10 Hz alpha plus noise")
plt.show()
```
