```python
import numpy as np
fs = 160
eeg = np.random.randn(4, fs * 10) * 10e-6   # 4 channels x 10 s, in volts
print(eeg.shape)            # (4, 1600)
print(eeg)
```

```python
first_second = eeg[:, :fs]  # all channels, first 160 samples
print(first_second.shape)   # (4, 160)
```

```python
ch_means = eeg.mean(axis=1) # one value per channel
print(ch_means.shape)       # (4,)
```

```python
eeg_uv = eeg * 1e6          # volts to microvolts, every element at once
print(eeg_uv)
```
